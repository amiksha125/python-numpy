# Advance Operation with Bussiness examples

```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
# data structure : [restaurant_id, 2021, 2022, 2023, 2024]
# restaurant id and their year-wise sales
sales_data = np.array( [
    [1, 150000, 180000, 220000, 250000], # Paradise Biryani
    [2, 120000, 140000, 160000, 190000], # Beijing Bites
    [3, 200000, 230000, 260000, 300000], # Pizza Hub
    [4, 180000, 210000, 240000, 270000],
    [5, 160000, 185000, 205000, 230000] # Chai Point
])

print("=== Zomato sales analysis ===")
print(sales_data)

print("Sales data shape ", sales_data.shape)
print("Sample data for first 3 restaurants \n", sales_data[ 0:3, :])
print("All years sales \n", sales_data[: , 1: ] )
```

```python
# total sales per year

print(np.sum (sales_data, axis = 0))
```

```python
# total sales of all years
yearly_total = np.sum(sales_data[ : , 1: ], axis = 0)
final_all_years = np.sum(yearly_total)
print("Total sales of each year ", yearly_total)
print("Total sales of all the years ", final_all_years)
```

```python
# Minimum sales per year
min_sales_year = np.min(sales_data[: , 1: ], axis = 0)
print(min_sales_year)

# Minimum sales per restaurant
min_sales_restaurant = np.min(sales_data[: , 1: ], axis = 1)
print(min_sales_restaurant)
```

```python
# Maximum sales per year
max_sales = np.max(sales_data[:, 1: ], axis = 0)
print(max_sales)

# Maximum sales per restaurant
max_sales_rest = np.max(sales_data[:, 1:], axis = 1)
print(max_sales_rest)
```

```python
# Average sales per restaurant
avg_sales = np.mean(sales_data[:, 1: ], axis=1)
print(avg_sales)

# Average sales of all the years
avg_all_years = np.mean(avg_sales)
print(avg_all_years)
```

```python
cumulative_sum  = np.cumsum(sales_data[:, 1:], axis = 1)
print(cumulative_sum)

plt.figure( figsize = (8, 6))
plt.plot(np.mean(cumulative_sum, axis = 0))
plt.title("Average cumulative sales per year")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
vector1 = np.array([1, 2, 3, 4, 5])
vector2 = np.array([6, 7, 8, 9, 10])

print("Vector addition ", vector1 + vector2)

print("Vector Multiplicatio ", vector1 * vector2)

print("Vectors Dot Multiplication ", np.dot(vector1, vector2))
```

```python
angle = np.arccos( np.dot(vector1, vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print(angle)
```

### Vectorized Operations

```python
restaurant_types = np.array( ['biryani', 'chinese', 'pizza', 'burger', 'cafe'])
vectorized_upper = np.vectorize(str.upper)
res_vectorized_upper = vectorized_upper(restaurant_types)
print(res_vectorized_upper)
```
